# Conhecendo a base

O diabetes é uma doença crônica grave na qual os indivíduos perdem a capacidade de regular efetivamente os níveis de glicose no sangue e pode levar a uma redução na qualidade de vida e na expectativa de vida.

O Sistema de Vigilância de Fatores de Risco Comportamentais (BRFSS) é uma pesquisa telefônica relacionada à saúde que é coletada anualmente pelo CDC (Centro de Controle e Prevenção de Doenças dos Estados Unidos). A cada ano, a pesquisa coleta respostas de milhares de americanos sobre comportamentos de risco relacionados à saúde, condições crônicas de saúde e o uso de serviços preventivos. Para este projeto, foi utilizado conjunto de dados disponível no Kaggle para o ano de 2015. 

https://www.kaggle.com/datasets/alexteboul/diabetes-health-indicators-dataset

```python
from src.config import DADOS_ORIGINAIS, DADOS_TRATADOS

import pandas as pd

pd.set_option('display.max_columns', None)

df_diabetes = pd.read_csv(DADOS_ORIGINAIS, compression='zip')

df_diabetes.head()
```

```python
df_diabetes.head()
```

## Verificando a estrutura da base

Nesta etapa, verificamos informações gerais das colunas, como tipos de dados e quantidade de registros.

**Por que fazer isso?**
Antes de tratar os dados, precisamos saber como eles estão armazenados e identificar possíveis problemas.

```python
df_diabetes.info()
```

## Traduzindo os nomes das colunas

Aqui trocamos os nomes originais das colunas por nomes em português e mais fáceis de entender.

traduzindo as colunas para facilitar o entendimento

```python
df_diabetes.columns = [
    "Diabetes",
    "PressaoAlta",
    "ColesterolAlto",
    "ColesterolExame",
    "IMC",
    "Fumante",
    "AVC",
    "ProblemaCardiaco",
    "AtividadeFisica",
    "ComeFrutas",
    "ComeLegumes",
    "ConsumoBebidaAlcoolica",
    "PlanoSaude",
    "SemDinheiroConsultas",
    "SaudeGeral",
    "DiasProblemasMentais",
    "DiasProblemasFisicos",
    "DificuldadeAndar",
    "Genero",
    "FaixaIdade",
    "Ensino",
    "FaixaRenda",
]
```

## Verificando os valores únicos

Essa informação ajuda a identificar quais colunas possuem apenas duas categorias e podem ser tratadas como variáveis binárias.

```python
df_diabetes.nunique()
```

## Identificando as colunas binárias

Aqui selecionamos as colunas que possuem exatamente dois valores diferentes.

**Por que fazer isso?**
Essas colunas representam situações com duas possibilidades e podem ser convertidas para categorias mais fáceis de interpretar.

```python
colunas_binarias = df_diabetes.nunique()[df_diabetes.nunique() == 2].index.to_list()

colunas_binarias
```

## Criando uma cópia para o tratamento

Criamos uma cópia da base original para realizar as alterações.

**Por que fazer isso?**
Assim mantemos os dados originais preservados e fazemos o tratamento em uma versão separada.

```python
df_diabetes_tratado = df_diabetes.copy()

df_diabetes_tratado.info()
```

## Transformando valores binários em categorias

Nesta etapa, os valores binários são transformados em categorias como **Sim** e **Não**. A coluna de gênero recebe categorias próprias.

**Por que fazer isso?**
Os valores ficam mais fáceis de entender e de interpretar durante a análise.

```python
for coluna in colunas_binarias:
    if coluna != 'Genero':
        df_diabetes_tratado[coluna] = pd.Categorical(df_diabetes_tratado[coluna]).rename_categories(['Não', 'Sim'])
    else:
        df_diabetes_tratado[coluna] = pd.Categorical(df_diabetes[coluna]).rename_categories(['Femenino', 'Masculino'])
```

Conferindo o resultado do tratamento

```python
df_diabetes_tratado.info()
```

## Transformando outras variáveis em categorias

Aqui organizamos variáveis como saúde geral, faixa de idade, ensino e renda em categorias ordenadas.

**Por que fazer isso?**
Essas informações possuem uma ordem natural. Manter essa ordem ajuda o Python a representar corretamente a relação entre as categorias.

```python
df_diabetes_tratado['SaudeGeral'] = pd.Categorical(
    df_diabetes_tratado['SaudeGeral'],
    ordered=True
).rename_categories(['Excelente', 'Muito boa', 'Boa', 'Aceitavel', 'Ruim'])

df_diabetes_tratado["FaixaIdade"] = pd.Categorical(
    df_diabetes_tratado["FaixaIdade"],
    ordered=True
).rename_categories(
    [
        "18-24",
        "25-29",
        "30-34",
        "35-39",
        "40-44",
        "45-49",
        "50-54",
        "55-59",
        "60-64",
        "65-69",
        "70-74",
        "75-79",
        "80+",
    ]
)

df_diabetes_tratado["Ensino"] = pd.Categorical(
    df_diabetes_tratado["Ensino"],
    ordered=True
).rename_categories(
    [
        "Sem estudo",
        "Primário",
        "Secundário inc.",
        "Secundário",
        "Faculdade inc. ou Técnico",
        "Faculdade +",
    ]
)

df_diabetes_tratado["FaixaRenda"] = pd.Categorical(
    df_diabetes_tratado["FaixaRenda"],
    ordered=True
).rename_categories(
    [
        "< $10.000",
        "$10.000-$14.999",
        "$15.000-$19.999",
        "$20.000-$24.999",
        "$25.000-$34.999",
        "$35.000-$49.999",
        "$50.000-$74.999",
        "$75.000+",
    ]
)

df_diabetes_tratado.head()
```

## Resumindo os dados numéricos

Nesta etapa, observamos um resumo estatístico das variáveis numéricas.

```python
df_diabetes_tratado.describe()
```

Aqui verificamos se os valores da coluna de dias de problemas físicos estão representados como números inteiros.

```python
df_diabetes_tratado['DiasProblemasFisicos'].apply(float.is_integer).all()
```

Identificando as colunas numéricas

```python
coluna_numericas = df_diabetes_tratado.select_dtypes(include='number').columns.to_list()

coluna_numericas
```

## Ajustando os tipos numéricos

Aqui convertemos as colunas numéricas para um formato inteiro mais adequado e econômico.

**Por que fazer isso?**
Isso deixa os dados mais organizados e pode reduzir o espaço utilizado pela base.

```python
for coluna in coluna_numericas:
    df_diabetes_tratado[coluna] = pd.to_numeric(
        df_diabetes_tratado[coluna],
        downcast='integer'
    )

df_diabetes_tratado.info()
```

## Salvando a base tratada

```python
df_diabetes_tratado.to_parquet(DADOS_TRATADOS, index=False)
df_diabetes.info()
```
